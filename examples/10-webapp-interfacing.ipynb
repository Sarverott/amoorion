# Webapp interfacing

Python web apps call the same object. `examples/spaces/` holds a Gradio and a
Streamlit app, ready to be Hugging Face Spaces, on `dryrun` only: a public demo must
never hold real accounts.

What both show is one function:

````python
from amoorion import Amoorion

# One object behind the whole app; use a real env file in production.
amo = Amoorion().use("dryrun")


# What a Gradio or Streamlit button would call: post, then list for its table.
def post(text):

    if text.strip():
        amo.publish(text=text)

    return [[message["id"], message["text"]] for message in amo.read(limit=20)]


# Called directly here, so that this notebook needs no web server.
post("from a web app")
````

The Gradio app around it:

```python
demo = gr.Interface(fn=post, inputs=gr.Textbox(label="Post"), outputs=gr.Dataframe(headers=["id", "text"]))
```
