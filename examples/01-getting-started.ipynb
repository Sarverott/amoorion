# Getting started

amoorion posts once to many platforms, reads them and answers from one place.
From Python it is one object. `dryrun` is a platform that sends nothing and only
remembers, so nothing here reaches anybody.

Create it, say which platforms to work on, publish:

```python
# Amoorion is one setup as an object, Plugins are the installed platforms.
from amoorion import Amoorion, Plugins

# No env file: a sandbox of its own. `use` picks the platforms to work on;
# dryrun is a platform that sends nothing, so it needs no account.
amo = Amoorion().use(Plugins["dryrun"])

# Every method is callable by its name. The answer has one result per platform.
amo.publish(
    title="Hello",
    text="first post from amoorion",
    tags=["amoorion"],
)

# is there a jupyter presentable ploter like matplotlib for mermaid? or just matplotlib if it fits with some plotting
# or just pandas
```

Everything amoorion can do is a method, and every method is callable by its name:

```python
# `read` answers a list; `limit=1` asks only for the newest message.
[latest] = amo.read(limit=1)


# Title and text became one text, as on most platforms.
print(latest["text"])

# A reply needs only the id of what it answers.
amo.reply(to=latest["id"], text="and a reply")["reply_to"]

# ugly
# wanna try prettyprint?

import pprint
```

What `read` fetched stays in the inbox, across platforms:

```python
# What was read is kept in the inbox, so this asks no platform at all.
[message["text"] for message in amo.inbox()]
```
